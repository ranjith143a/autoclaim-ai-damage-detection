### 3. Visual Performance Comparison


```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Validation RMSE
sns.barplot(x='Val_RMSE_INR', y='Algorithm', data=df_results, palette='Blues_r', ax=axes[0])
axes[0].set_title('Validation RMSE in INR (Lower is Better)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('RMSE (INR)')

# Validation R2
sns.barplot(x='Val_R2', y='Algorithm', data=df_results, palette='Greens_r', ax=axes[1])
axes[1].set_title('Validation R² Score (Higher is Better)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('R² Score')

plt.tight_layout()
plt.show()

```

### 4. Key Takeaways
1. **Gradient Boosting Dominance:** The top 4 performing algorithms are all gradient boosted decision trees:
   - Gradient Boosting: Val RMSE ₹14,368, R² = 0.8327
   - LightGBM: Val RMSE ₹14,391, R² = 0.8321
   - HistGradientBoosting: Val RMSE ₹14,393, R² = 0.8321
   - XGBoost: Val RMSE ₹14,499, R² = 0.8296
2. **Computational Speed:** LightGBM and HistGradientBoosting train in under 0.4 seconds, making them ideal candidates for hyperparameter tuning.
3. **Linear Models Ceiling:** Linear, Ridge, and Lasso achieve $R^2 \approx 0.797$ and MAE $\approx$ ₹10,436, indicating strong baseline linearity but inability to capture complex non-linear part-interaction terms.
